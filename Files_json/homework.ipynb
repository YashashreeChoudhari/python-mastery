Q. pathlib

```python
from pathlib import Path
```

```python
folder=Path("path_home")
folder.mkdir(exist_ok=True)
```

```python
file = folder/"path_home.txt"
file.write_text("Hi homework file",encoding="utf-8")
```

```python
parent_folder = Path("path_home")
txt_files = parent_folder.glob("*.txt")
```

```python
list(txt_files)
```

```python
base_folder = Path("path_home").resolve()
file_name = input("Enter the file:")
file_path = (base_folder/ file_name).resolve()

if file_path.is_relative_to(base_folder):
    data = file_path.read_text()
    print(data)
else:
    print("access denied")
```

Q. JSON

```python
import json

dict={
    "name":"Siya",
    "age":25,
    "employee":True
}

data=json.dumps(dict,indent=2)
print(data)
```

```python
place={
    "city":"Mumbai",
    "pin":400001
}

with open("place.json","w") as f:
    json.dump(place, f, indent=2)
```

```python
with open("place.json", "r") as f:
    data_place = json.load(f)
```

```python
data_place["city"]
```

```python
contact= [
    {
        "name": "Rahul",
        "phone": "9876543210",
        "email": "rahul@gmail.com"
    },
    {
        "name": "Priya",
        "phone": "9123456780",
        "email": "priya@gmail.com"
    },
    {
        "name": "Amit",
        "phone": "9988776655",
        "email": "amit@gmail.com"
    }
]

with open("contact.json", "w") as file:
    json.dump(contact, file, indent=2)

# Reload contacts from JSON file
with open("contact.json", "r") as file:
    loaded_contacts = json.load(file)


for contact in loaded_contacts:
    print("Name:", contact["name"])
    print("Phone:", contact["phone"])
    print("Email:", contact["email"])
    print("--------------------")
```
