```python
#open function  - used to open file
#write function  - used to write inside file
with open("my_first_file.txt","w") as f:
    text = "Hello"
    f.write(text)
```

```python
#read - used to write
with open("my_first_file.txt","r") as f:
    data = f.read()
print(data)
```

```python
with open("my_first_file.txt","r") as f:
    data = f.read()
    for line in data.split("\n"):
        print(line)
```

```python
with open("my_first_file.txt","r") as f:
    for line in f:
        print(line.strip())
```

```python
#chat history project
def create_history()->None:
    with open("agent_chat_history.json", "w") as f:
        json.dump([], f, indent=4)


def add_history(role: str, content: str)->None:
    with open("agent_chat_history.json", "r") as f:
        history = json.load(f)

    history.append({"role": role, "content": content})

    with open("agent_chat_history.json", "w") as f:
        json.dump(history, f, indent=4)


def read_history()->list:
    with open("agent_chat_history.json", "r") as f:
        print(json.load(f))
```

```python
create_history()
```

```python
add_history("user", "Hi, how are you")
```

```python
add_history("system", "Act as developer")
```
