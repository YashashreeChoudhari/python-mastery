Challenge: Rotate a List

```python
def rotate_list(numbers, k):
    return numbers[-k:] + numbers[:-k]
        
rotate_list([1, 2, 3, 4, 5], 2)
```

Strings

Concept: String processing + loops
Challenge: Compress a String

```python
def compress_string(text):
    new = ""
    current = text[0]
    count = 0

    for i in text:
        if i == current:
            count += 1
        else:
            new += current + str(count)
            current = i
            count = 1

    new += current + str(count)
    return new

compress_string("aaabbc")
```

Lists

Find the Longest Consecutive Sequence

```python
def longest_consecutive(numbers):
    longest=0

    for number in numbers:
        current=number
        count=1

        while current+1 in numbers:
            current+=1
            count+=1

        if count>longest:
            longest=count

    return longest
    
longest_consecutive([100, 4, 200, 1, 3, 2])
```

Merge Two Sorted Lists

```python
def merge_sorted(list1, list2):
    merge=list1+list2
    return sorted(merge)
    
merge_sorted([1, 3, 5], [2, 4, 6])
```

```python
def merge_sorted(list1, list2):
    merge = []
    i=0
    j=0

    while i<len(list1) and j<len(list2):
        if list1[i]<list2[j]:
            merge.append(list1[i])
            i += 1
        else:
            merge.append(list2[j])
            j += 1
            
    while i < len(list1):
        merge.append(list1[i])
        i += 1

    while j < len(list2):
        merge.append(list2[j])
        j += 1
    return merge
merge_sorted([1, 3, 5], [2, 4, 6])
```

Dictionaries + Lists
Challenge: Invert a Dictionary

```python
def invert_dict(data):
    result = {}

    for key in data:
        result[data[key]]=key

    return result

invert_dict({
    "name": "John",
    "age": 25,
    "city": "Delhi"
})
```
