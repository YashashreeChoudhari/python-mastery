Then write Python code that:

Opens the file.

Removes the newline character from each line.
Stores the names in a list.
Prints the list.

```python
with open("name.txt","r") as f:
    context=f.read()
    
    print(context)
```

```python
with open("name.txt","r") as file:
    names = []

    for line in file:
        names.append(line.strip())

    file.close()

print(names)
```

Then write Python code that reads the file and calculates the total without using sum().

```python
with open("numbers.txt","w")as f:
    f.write("10\n")
    f.write("20\n")
    f.write("30\n")
    f.write("40\n")
    f.write("50\n")
```

```python
with open("numbers.txt", "r") as f:
    total = 0

    for line in f:
        number = int(line.strip())
        total = total + number

print(total)
```

File writing + loop

Write the squares of numbers 1 through 5, one per line:


```python
no = [1, 2, 3, 4, 5]

with open("squares.txt", "w") as f:
    for i in no:
        f.write(str(i) + "\n")
```

```python
with open("squares.txt", "w") as f:
    for i in range(1, 6):
        f.write(str(i * i) + "\n")

with open("squares.txt", "r") as f:
    for line in f:
        print(line.strip())
```
