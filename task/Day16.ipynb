Challenge: Word Frequency Counter

```python
def word_frequency(text):
    result = {}

    for word in text.lower().split():
        word = word.strip(".,!?")
        if word in result:
            result[word] += 1
        else:
            result[word] = 1

    return result

text = "Python, is easy and python! is powerful. PYTHON"
print(word_frequency(text))
```

Lists + Functions

```python
def second_largest(numbers):
    larg=numbers[0]
    sec_large=numbers[0]
    
    for i in numbers:
        if i>larg:
            sec_large=larg
            larg=i
        elif i>sec_large and i != larg:
            sec_large=i
    return sec_large
    
second_largest([10, 5, 8, 20, 15])
```
