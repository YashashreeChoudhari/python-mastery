Find the Largest Difference

```python
def largest_difference(numbers):
    large=numbers[0]
    small=numbers[0]
    
    for i in numbers:
        if large<i:
            large=i
        elif i<small:
            small=i
    
    return large-small
    
largest_difference([10, 3, 7, 2, 15])
```

Move Negative Numbers to the Left

```python
def move_negatives(numbers):
    negative=[]
    rest_list=[]
    
    for i in numbers:
        if i<0:
            negative.append(i)
        else:
            rest_list.append(i)
    return negative+rest_list

move_negatives([3, -1, 4, -2, 0, -5, 7])
```

Check Balanced Parentheses

```python
def is_balanced(text):
    count = 0

    for i in text:
        if i == "(":
            count += 1
        elif i == ")":
            count -= 1

            if count < 0:
                return False

    return count == 0

is_balanced("(hello)")
is_balanced("((hello))")
is_balanced(")( ")
```

Remove Consecutive Duplicate Characters

```python
def remove_consecutive_duplicates(text):
    result=""
    previous=""

    for char in text:
        if char!=previous:
            result+=char

        previous=char

    return result
    
    
remove_consecutive_duplicates("aaabbc")
```
