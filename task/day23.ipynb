Product of Array Except Self

```python
def product_except_self(numbers):
    result = []

    for i in range(len(numbers)):
        product = 1

        for j in range(len(numbers)):
            if i != j:
                product *= numbers[j]

        result.append(product)

    return result


product_except_self([1, 2, 3, 4])
```

```python
def product_except_self(numbers):
    result = [1] * len(numbers)

    prefix = 1

    for i in range(len(numbers)):
        result[i] = prefix
        prefix *= numbers[i]

    suffix = 1

    for i in range(len(numbers) - 1, -1, -1):
        result[i] *= suffix
        suffix *= numbers[i]

    return result


print(product_except_self([1, 2, 3, 4]))
print(product_except_self([2, 3, 4]))
```

Recursion

Write a program to calculate the factorial of a number using recursion.

```python
def factorial(num):
    fact=1
    while num>0:
        fact=fact*num
        num=num-1
        
    return fact
factorial(5)
```

```python
def factorial(num):
    if num == 0:
        return 1

    return num * factorial(num - 1)


print(factorial(5))
```

Dictionaries + lists + nested data

```python
students = {
    "Alice": [80, 90, 70],
    "Bob": [60, 75, 85],
    "Charlie": [90, 95, 100]
}

averages = {}

for name in students:
    total = 0
    count = 0

    for grade in students[name]:
        total = total + grade
        count = count + 1

    average = total / count
    averages[name] = round(average, 2)

print(averages)
```

```python
students = {
    "Alice": [80, 90, 70],
    "Bob": [60, 75, 85],
    "Charlie": [90, 95, 100]
}

result = {}

for name,grades in students.items():
    total = 0

    for i in grades:
        total+=i
        
    average=total/len(grades)
    result[name]=round(average,2)

print(result)
```

zip()

Given:

names = ["Alice", "Bob", "Charlie"]
scores = [85, 72, 91]

Create this dictionary:

{
    "Alice": 85,
    "Bob": 72,
    "Charlie": 91
}

```python
names = ["Alice", "Bob", "Charlie"]
scores = [85, 72, 91]
dict={}

for name,score in zip(names,scores):
    dict[name]=score
    
print(dict)
```
