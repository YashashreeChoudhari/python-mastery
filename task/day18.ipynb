Find the Missing Number

```python
def find_missing(numbers):
    for i in range(1,len(numbers)+2):
        if i not in numbers:
            return i
        
find_missing([1, 2, 3, 5])
```

First Non-Repeating Character

```python
def first_unique_char(text):
    unique={}
    for i in text.lower():
        if i!=" ":
            if i in unique:
                unique[i]+=1
            else:
                unique[i]=1
                
    for i in text.lower(): 
        if i!=" ":     
            if unique[i]==1:
                return i
    
    return None

# first_unique_char("aabbcdd")
# first_unique_char("swiss")
first_unique_char("a b")
first_unique_char("aabbcc")
```

```python
def first_unique_char(text):
    unique = {}
    text = text.lower()

    # Pass 1: count characters
    for i in text:
        if i != " ":
            if i in unique:
                unique[i] += 1
            else:
                unique[i] = 1

    # Pass 2: find first character appearing once
    for i in text:
        if i != " " and unique[i] == 1:
            return i

    return None


print(first_unique_char("aabbcdd"))  # c
print(first_unique_char("swiss"))    # w
print(first_unique_char("a b"))      # a
print(first_unique_char("aabbcc"))   # None
```

Count Occurrences of Each Number

```python
def count_numbers(numbers):
    freq={}
    
    for i in numbers:
        if i in freq:
            freq[i]+=1
        else:
            freq[i]=1
            
    return freq

count_numbers([1, 2, 2, 3, 1, 2, 4])
```

Lists + Functions

```python
def find_common(numbers1, numbers2):
    common=[]
    for i in numbers1:
        if i in numbers2 and i not in common:
            common.append(i)
    return common
            
find_common([1, 2, 2, 3, 4], [2, 2, 4, 5])
```

Lists + Strings

Concept: Nested loops, string comparison
Challenge: Find Anagrams

```python
def are_anagrams(word1, word2):
    count1 = {}
    count2 = {}

    for i in word1.lower():
        if i!=" ":
            if i in count1:
                count1[i] += 1
            else:
                count1[i] = 1

    for i in word2.lower():
        if i!=" ":
            if i in count2:
                count2[i] += 1
            else:
                count2[i] = 1

    #REMEMBER dict compare using key and value not order
    return count1 == count2  

are_anagrams("listen", "silent")
are_anagrams("conversation", "voices rant on")
```

Move Zeros to the End

```python
def move_zeros(numbers):
    not_zero=[]
    zero=[]
    final=[]
    for i in numbers:
        if i!=0:
            not_zero.append(i)
        else:
            zero.append(i)
        
    final=not_zero+zero    
    return final
            
move_zeros([0, 1, 0, 3, 12])
```
