Challenge: Find Pairs That Add to a Target

```python
#duplicate output
def find_pairs(numbers, target):
    pairs = []

    for i in numbers:
        for j in numbers:
            if i + j == target:
                pairs.append([i, j])

    return pairs

find_pairs([1, 2, 3, 4, 5], 6)
```

```python
#not applicable for same numbers in list like [3, 3, 3]
def find_pairs(numbers, target):
    pairs = []

    for i in numbers:
        for j in numbers:
            if i + j == target and i < j:
                pairs.append([i, j])

    return pairs

find_pairs([1, 2, 3, 4, 5], 6)
find_pairs([3, 3, 3], 6)
```

```python
def find_pairs(numbers, target):
    pairs = []

    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] + numbers[j] == target:
                pairs.append([numbers[i], numbers[j]])

    return pairs

print(find_pairs([1, 2, 3, 4, 5], 6))
find_pairs([3, 3, 3], 6)
```

Challenge: Reverse the Order of Words

```python
def reverse_words(text):
    rev=""
    
    for i in text.split():
        rev=i+" "+rev
    
    return rev.strip()

print(reverse_words("I love Python"))
```

```python
def reverse_words(text):
    rev = []
    
    for i in text.split():
        rev.insert(0, i)
    
    return " ".join(rev)

print(reverse_words("I love Python"))
```

Find the Second Smallest Unique Number

```python
def second_smallest(numbers):
    smallest=numbers[0]
    sec_smallest=numbers[0]
    
    for i in numbers:
        if i<smallest:
            sec_smallest=smallest
            smallest=i
        elif i<sec_smallest and i!=smallest:
            sec_smallest=i
    
    return sec_smallest
            
    
second_smallest([10, 5, 8, 20, 15])
second_smallest([4, 1, 2, 1, 3])
```

```python
def second_smallest(numbers):
    smallest = numbers[0]
    sec_smallest = float('inf')

    for i in numbers:
        if i < smallest:
            sec_smallest = smallest
            smallest = i
        elif i < sec_smallest and i != smallest:
            sec_smallest = i

    return sec_smallest

print(second_smallest([10, 5, 8, 20, 15]))
second_smallest([4, 1, 2, 1, 3])
```

Challenge: Remove All Occurrences of a Value

```python
def remove_value(items, value):
    new=[]
    for i in items:
        if i!=value:
            new.append(i)
    return new

remove_value([1, 2, 3, 2, 4, 2], 2)
```

```python
def count_word_lengths(text):
    result = {}

    for word in text.split():
        length = len(word)

        if length in result:
            result[length] = result[length] + 1
        else:
            result[length] = 1

    return result
count_word_lengths("I love learning Python")
```
