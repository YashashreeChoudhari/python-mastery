Challenge: Remove Duplicates Without Using set()

```python
def remove_duplicates(numbers):
    new=[]
    for i in numbers:
        if i not in new:
            new.append(i)
    return new  
        
remove_duplicates([5, 5, 5, 2, 2, 8])
```

Dictionaries + Functions

Group Words by Their First Letter

```python
def group_by_first_letter(words):
    groups = {}

    for word in words:
        first = word[0]

        if first in groups:
            groups[first].append(word)
        else:
            groups[first] = [word]

    return groups


group_by_first_letter(
    ["apple", "banana", "apricot", "blueberry", "cherry", "avocado"])
```
