Context Manager-run setup code on entry and guranteed cleanup code on exit

```python
from contextlib import contextmanager
```

```python
import time

@contextmanager  #decorator

def timer(label):
    #start
    start_time = time.time()
    print({label})
    print("starting time is", start_time)
    
    yield
    
    #end
    end_time = time.time()
    print(f"{label} took {end_time - start_time}")

with timer("my task"):
    total = sum(range(500))
    print("Total sum is", total)
```

```python
#methods
from dataclasses import dataclass, field

@dataclass
class Agent:
    name: str
    model: str = 'gpt.5'
    tools: list = field(default_factory=list)

    def add_tool(self, tool):
        self.tools.append(tool)
```

Protocol - rules that should be followed

```python
from typing import Protocol
```

```python
class Runnable(Protocol):
    def run(self) -> str:...

def execute(tool: Runnable) -> str:
    return tool.run()
```

```python
class SearchTool:
    def run(self) -> str:
        return "searching"
```

```python
tool=SearchTool()
execute(tool)
```

```python
class SearchToolWithoutRun:
    def ran(self) -> str:
        return "searching"
```

```python
bad_tool = SearchToolWithoutRun()
execute(bad_tool)
```
