```python
@contextmanager 
def context_printing(label):
    print("Enter")
    
    yield
    print("Exit")

with context_printing("print"):
    print("middle")
```

```python
from collections import defaultdict
```

```python
students = [
    ("Meena", "A"),
    ("Siya", "B"),
    ("Riya", "A"),
    ("Gita", "C"),
    ("Tina", "B")
]

group=defaultdict(list)

for name, grade in students:
    group[grade].append(name)

group
```

```python
from functools import lru_cache
```

```python
@lru_cache()
def fib(n):
    print("calculating")
    
    if n <= 1:
        return n
    
    return fib(n - 1) + fib(n - 2)
```

```python
fib(10)
```

```python
fib(10)
```

```python
fib(5)
```

```python
fib(11)
```

Typing deep

```python
def count_words(words: list[str]) -> dict[str, int]:
    result = {}

    for word in words:
        result[word] = result.get(word, 0) + 1

    return result

count_words("hi i m yashashree")
```

```python
def find_number(numbers: list[int], target: int) -> int | None:
    if target in numbers:
        return target
    return None
```

```python
find_number([1, 2, 3], 2)
```
