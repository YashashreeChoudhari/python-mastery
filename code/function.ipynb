```python
def calculate_total(numbers):
    total = 0

    for num in numbers:
        total += num

    return total

print(calculate_total([10, 20, 30]))
print(calculate_total([5, 15, 25]))
```

```python
def square(number):
    sq=number*number
    return sq

print(square(5))
```

```python
def is_even(number):
    if number%2==0:
        return True
    else:
        return False

print(is_even(10))  
print(is_even(7))   

#or

def is_even(number):
    return number % 2 == 0
print(is_even(10))  
```

```python
def calculate_total(price, quantity):
    total=price*quantity
    return total
print(calculate_total(500, 3))
```

```python
def find_max(numbers):
    max=numbers[0]
    for i in numbers:
        if i>max:
            max=i
    return max

print(find_max([10, 25, 7, 40, 15]))
print(find_max([-10, -5, -20]))
```

```python
def find_max_even(numbers):
    lar = None

    for i in numbers:
        if i % 2 == 0:
            if lar is None or i > lar:
                lar = i

    return lar

print(find_max_even([7, 12, 5, 20, 9, 14, 3]))
print(find_max_even([101, 20, 40]))
    
```

```python
#Function + Dictionary
students = {
    "Amit": 85,
    "Rahul": 72,
    "Priya": 91,
    "Neha": 78,
    "Vikram": 95
}

def find_top_student(students):
    top=0
    name=""
    
    for key,value in students.items():
        if value>top:
            top=value
            name=key
    return name,top

print(find_top_student(students))
```

```python
#Function + Dictionary + Condition
students = {
    "Amit": 85,
    "Rahul": 72,
    "Priya": 91,
    "Neha": 78,
    "Vikram": 95,
    "Karan": 88
}

def find_passed_students(students):
    name=[]
    
    for key,value in students.items():
        if value>=80:
            name.append(key)
    return name

print(find_passed_students(students))     
```

```python
products = {
    "Laptop": {"price": 50000, "quantity": 5},
    "Mouse": {"price": 800, "quantity": 20},
    "Keyboard": {"price": 1500, "quantity": 12},
    "Monitor": {"price": 12000, "quantity": 7},
    "Webcam": {"price": 3000, "quantity": 15}
}

def calculate_revenues(products):
    product_name=""
    highest_revenue=0
    
    for key,value in products.items():
        price=value["price"]
        quantity=value["quantity"]
        revenue=price*quantity
        
        if revenue>highest_revenue:
            highest_revenue=revenue
            product_name=key
    return highest_revenue,product_name

print(calculate_revenues(products))
```
