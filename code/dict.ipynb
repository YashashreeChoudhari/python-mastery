```python
# Create this dictionary
names = ["Amit", "Rahul", "Priya"]
marks = [85, 72, 91]

score = {}

for i in range(len(names)):
    score[names[i]] = marks[i]

print(score)

# or

scores = {}

for name, mark in zip(names, marks):
    scores[name] = mark
```

```python
#Looping Through a Nested Dictionary

employees = {
    "Amit": {
        "role": "Backend",
        "salary": 75000
    },
    "Rahul": {
        "role": "Frontend",
        "salary": 65000
    }
}

for name, details in employees.items():
    print(name)
    print(details)

for name, details in employees.items():
    print(name, details["role"], details["salary"])
    
for name, details in employees.items():
    print(name)

    for key, value in details.items():
        print(key, value)
```

```python
# Create a dictionary containing the frequency of each number.
numbers = [10, 20, 10, 30, 20, 10, 40, 30, 50, 20]
count = {}

for i in range(len(numbers)):
    if numbers[i] in count:
        count[numbers[i]] += 1
    else:
        count[numbers[i]] = 1

print(count)
    
```

```python
numbers = [10, 20, 10, 30, 20, 10, 40, 30, 20, 20]
freq = {}

# Step 1: Count frequency
for num in numbers:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

# Step 2: Find the highest frequency
highest = 0
most_frequent = 0

for num, count in freq.items():
    if count > highest:
        highest = count
        most_frequent = num

print(most_frequent)
```

```python
products = ["Laptop", "Mouse", "Keyboard", "Monitor", "Webcam"]
sales = [5, 20, 12, 7, 15]
prices = [50000, 800, 1500, 12000, 3000]

revenues = []
# Each product's revenue
for i in range(len(products)):
    revenue = sales[i] * prices[i]
    revenues.append(revenue)
    print(products[i], ":", revenue)
    
# The product with the highest revenue
highest_revenue=max(revenues)
highest_index = revenues.index(highest_revenue)

# The highest revenue
print("Product with highest revenue:", products[highest_index])
print("Highest revenue:", highest_revenue)
```

```python
users = {
    "Amit": {"Python", "SQL", "Git"},
    "Rahul": {"Java", "Git"},
    "Priya": {"Python", "Git"},
    "Neha": {"Python", "Docker"}
}

active_users = {"Amit", "Priya", "Neha"}
unique = set()

for user in active_users:
    for skill in users[user]:
        unique.add(skill)

print(unique)
```
