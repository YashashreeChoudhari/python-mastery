```python
employee = {
    "name": "Rahul",
    "role": "Backend",
    "salary": 75000,
    "experience": 3
}
#items(both key and value)
for key,value in employee.items():
    print(f"{key} -> {value}")

# Keys
for key in employee:
    print(key)

# Values
for value in employee.values():
    print(value)
```

```python
student = {
    "name": "Amit",
    "age": 21,
    "marks": 85,
    "course": "Python"
}

# Remove "age" using pop()
student.pop("age")

# Remove "course" using del
del student["course"]

# Print the remaining dictionary
print(student)

# Use clear() and print it again
student.clear()
print(student)
```

```python
scores = {
    "Amit": 85,
    "Rahul": 72,
    "Priya": 91,
    "Neha": 68
}

for name in scores:
    print(name)

for marks in scores.values():
    print(marks)

for name, marks in scores.items():
    print(name, marks)

for name, marks in scores.items():
    if marks>=80:
        print(name, marks)
```

```python
# Create this dictionary
names = ["Amit", "Rahul", "Priya"]
marks = [85, 72, 91]

score = {}

for i in range(len(names)):
    score[names[i]] = marks[i]

print(score)

# or

scores = {}

for name, mark in zip(names, marks):
    scores[name] = mark
```

```python
#dictionary + counting
# value → frequency

languages = ["Python", "Java", "Python", "C++", "Java", "Python"]

freq = {}
for language in languages:
    if language in freq:
        freq[language] += 1
    else:
        freq[language] = 1 
        
print(freq) 
```

```python
numbers = [10, 20, 10, 30, 20, 10, 40, 30]
count={}

for num in numbers:
    if num in count:
        count[num]+=1
    else:
        count[num]=1
print(count)
```

```python
developers = [
    ("Amit", "Python"),
    ("Rahul", "Java"),
    ("Amit", "SQL"),
    ("Rahul", "Git")
]

skills = {}

for name, skill in developers:
    skills.setdefault(name, [])
    skills[name].append(skill)

print(skills)
```

```python
employees = [
    ("Amit", "Backend"),
    ("Rahul", "Frontend"),
    ("Amit", "DevOps"),
    ("Priya", "Backend"),
    ("Rahul", "React"),
    ("Priya", "Python")
]

dict={}

for name,feild in employees:
    dict.setdefault(name,[])
    dict[name].append(feild)

print(dict)
```
