# Lambda

```python
add = lambda a, b: a + b
print(add(3, 5))
```

```python
#lamba double
double=lambda a: a*2
print(double(7))
```

```python
is_even=lambda a:a%2==0
is_odd=lambda a:a%2!=0

print(is_even(8))
print(is_even(7))
```

```python
get_length = lambda a:len(a)
print(get_length("Python"))
print(get_length("Hello World"))
```

# Map() 

```python
# Use map() + lambda to add 5 to every number.
numbers = [2, 4, 6, 8, 10]
result=list(map(lambda a: a+5,numbers))
print(result)
```

```python
#square
numbers = [1, 2, 3, 4, 5]
result=list(map(lambda a: a*a,numbers))
print(result)
```

```python
#string
names = ["alice", "bob", "charlie", "david"]
name_result=list(map(lambda a:a.upper(),names))
print(name_result)
```

# Filter()

```python
numbers = [3, 8, 11, 14, 17, 20, 25]
result=list(filter(lambda a: a>10,numbers))
print(result)
```

```python
#string
words = ["cat", "elephant", "dog", "tiger", "ant"]
result=list(filter(lambda a: len(a)>3,words))
print(result)
```

```python
numbers = [4, 7, 10, 13, 16, 19, 22]
result=list(filter(lambda a: a%2==0 and a>10,numbers))
print(result)
```

# map and filter

```python
numbers = [1, 2, 3, 4, 5, 6]
fil_result=list(filter(lambda a: a%2==0,numbers))
final_res=list(map(lambda a: a * a, fil_result))
print(final_res)

#or

numbers = [1, 2, 3, 4, 5, 6]

final_res = list(
    map(
        lambda a: a * a,
        filter(lambda a: a % 2 == 0, numbers)
    )
)

print(final_res)
```

```python
numbers = [2, 5, 8, 11, 14, 17, 20]
result=list(
    map(
        lambda a: a*3,
        filter(lambda a: a>10,numbers)
    )
)
print(result)
```

# sorted()

```python
words = ["apple", "fig", "banana", "kiwi"]
result = sorted(words, key=lambda x: len(x))
print(result)
```

```python
numbers = [-10, 3, -2, 8, -5]
result = sorted(numbers, key=lambda x: abs(x))
print(result)
```
