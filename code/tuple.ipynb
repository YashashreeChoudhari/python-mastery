```python
developers = [
    ("Amit", "Backend"),
    ("Rahul", "Frontend"),
    ("Priya", "Backend"),
    ("Neha", "DevOps"),
    ("Vikram", "Frontend")
]
dev_set=set(developers)
active = {"Amit", "Priya", "Neha", "Vikram"}

# 1. Developers who are active
for i in active:
    print(i)
    
# Developers who are not active
for i in developers:
    if i[0] not in active:
        print(i[0])

# Unique roles currently represented by active developers
unique=set()
for i in developers:
    if i[0] in active:
        unique.add(i[1])

print(unique)

#Which roles have at least one inactive developer?
inactive_roles = set()

for name, role in developers:
    if name not in active:
        inactive_roles.add(role)

print(inactive_roles)
```

```python
employees = [
    ("Amit", "Backend"),
    ("Rahul", "Frontend"),
    ("Priya", "Backend"),
    ("Neha", "DevOps"),
    ("Vikram", "Frontend"),
    ("Karan", "Backend")
]
project_a = {"Amit", "Priya", "Karan", "Rahul"}
project_b = {"Priya", "Neha", "Vikram"}

# Employees working on both projects
print(project_a & project_b)

# Employees working only on Project A
print(project_a - project_b)

# Employees working only on Project B
print(project_b - project_a)

#Employees not working on either project
c=project_a | project_b

for name,role in employees:
    if name not in c:
        print(name)

#Unique roles represented by employees working on either project
uni_role=set()
d= project_a ^ project_b

for name,role in employees:
    if name in d:
        uni_role.add(role)
        
print(uni_role)
```

```python
developer1 = {"Python", "SQL", "Git", "Docker"}
developer2 = {"Python", "Java", "Git", "AWS"}

# Find:
# Skills both developers have
print(developer1 & developer2)

# Skills only developer 1 has
print(developer1 - developer2)

# Skills only developer 2 has
print(developer2 - developer1)

# All unique skills
print(set(developer1 | developer2))
```

```python
usernames = [
    "amit",
    "rahul",
    "priya",
    "amit",
    "neha",
    "rahul",
    "vikram",
    "priya"
]
print(set(usernames))

#dupli
dupli=[]
new=[]
for i in usernames:
    if i in new:
        dupli.append(i)
    else:
        new.append(i)

print(dupli)

#dupli count
dupli=[]
new=[]
count=0
for i in usernames:
    if i in new:
        dupli.append(i)
        count+=1
    else:
        new.append(i)
print(count)
```
